```python
positive_words = [
    'gain', 'gains', 'gained', 'gainer', 'gainers',
    'growth', 'grew', 'grow', 'growing',
    'profit', 'profits', 'profitable', 'profitability',
    'rise', 'rises', 'rising', 'risen', 'rose',
    'rally', 'rallied', 'rallying', 'rallies',
    'surge', 'surges', 'surged', 'surging',
    'bullish', 'bull', 'bull run',
    'record', 'record-high', 'all-time high',
    'upgrade', 'upgraded', 'upgrades',
    'buyback', 'share buyback', 'repurchase',
    'loan growth', 'credit growth',
    'oversubscribed', 'oversubscription',
    'multibagger', 'broad-based rally'
]
```

```python
negative_words = [
    'loss', 'losses', 'lost', 'lose', 'losing',
    'decline', 'declines', 'declined', 'declining',
    'crash', 'crashes', 'crashed', 'crashing',
    'plunge', 'plummet', 'nosedive', 'collapse',
    'bearish', 'bear market',
    'downgrade', 'downgraded', 'downgrades',
    'recession', 'stagflation', 'slowdown',
    'layoffs', 'retrenchment', 'downsizing',
    'default', 'bankruptcy', 'insolvency',
    'npa', 'bad loan', 'write-off',
    'rate hike', 'hawkish', 'monetary tightening',
    'fii selling', 'dii selling',
    'institutional selling',
    'circuit breaker', 'lower circuit'
]
```

```python
def sentiment_score(text):
    text = str(text).lower()

    score = 0

    for word in positive_words:
        score += text.count(word)

    for word in negative_words:
        score -= text.count(word)

    return score
```

```python
merged['sentiment_score'] = merged['Title_y'].apply(sentiment_score)
```

```python
import pandas as pd
import numpy as np

ml_df = pd.read_csv(
    "final_project_dataset.csv",
    parse_dates=['date']
)

ml_df = ml_df.sort_values('date').reset_index(drop=True)

print(ml_df.shape)
display(ml_df.head())
```

```python
ml_df['next_day_target'] = ml_df['target'].shift(-1)
```

```python
ml_df = ml_df.dropna(
    subset=['next_day_target']
)

ml_df['next_day_target'] = (
    ml_df['next_day_target'].astype(int)
)

print(
    ml_df[
        ['date', 'target', 'next_day_target']
    ].head(10)
)
```

```python
ml_df['sentiment_ma3'] = (
    ml_df['sentiment_score'].rolling(3).mean()
)

ml_df['sentiment_ma7'] = (
    ml_df['sentiment_score'].rolling(7).mean()
)

ml_df['sentiment_std3'] = (
    ml_df['sentiment_score'].rolling(3).std()
)
```

```python
ml_df['movement_lag1'] = (
    ml_df['movement'].shift(1)
)

ml_df['movement_lag2'] = (
    ml_df['movement'].shift(2)
)
```

```python
ml_df['volume_change'] = (
    ml_df['shares_traded'].pct_change()
)
```

```python
ml_df['high_low_spread'] = (
    (ml_df['high'] - ml_df['low'])
    / ml_df['open']
    * 100
)
```

```python
ml_df = ml_df.dropna().reset_index(drop=True)

print("ML dataset shape:", ml_df.shape)
```

```python
feature_cols = [
    'sentiment_score',
    'sentiment_ma3',
    'sentiment_ma7',
    'sentiment_std3',
    'movement_lag1',
    'movement_lag2',
    'volume_change',
    'high_low_spread'
]

print(ml_df[feature_cols].head())
```

```python
X = ml_df[feature_cols]
y = ml_df['next_day_target']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
split_idx = int(len(ml_df) * 0.8)

X_train = X.iloc[:split_idx]
X_test = X.iloc[split_idx:]

y_train = y.iloc[:split_idx]
y_test = y.iloc[split_idx:]

print("Training days:", len(X_train))
print("Testing days:", len(X_test))
```

```python
print(
    "Training period:",
    ml_df['date'].iloc[0],
    "to",
    ml_df['date'].iloc[split_idx - 1]
)

print(
    "Testing period:",
    ml_df['date'].iloc[split_idx],
    "to",
    ml_df['date'].iloc[-1]
)
```

```python
majority_baseline = (
    y_test == y_test.mode()[0]
).mean()

print(
    "Majority-class baseline:",
    round(majority_baseline * 100, 2),
    "%"
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
log_reg = LogisticRegression(
    max_iter=1000
)

log_reg.fit(
    X_train_scaled,
    y_train
)
```

```python
y_pred_lr = log_reg.predict(
    X_test_scaled
)
```

```python
lr_accuracy = accuracy_score(
    y_test,
    y_pred_lr
)

print(
    "Logistic Regression Accuracy:",
    round(lr_accuracy * 100, 2),
    "%"
)

print(
    classification_report(
        y_test,
        y_pred_lr
    )
)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=5,
    min_samples_leaf=10,
    random_state=42
)

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

rf_accuracy = accuracy_score(
    y_test,
    y_pred_rf
)

print(
    "Random Forest Accuracy:",
    round(rf_accuracy * 100, 2),
    "%"
)
```

```python
results = pd.DataFrame({
    'Model': [
        'Majority Baseline',
        'Logistic Regression',
        'Random Forest'
    ],
    'Accuracy': [
        majority_baseline,
        lr_accuracy,
        rf_accuracy
    ]
})

results['Accuracy (%)'] = (
    results['Accuracy'] * 100
).round(2)

display(results)
```

```python
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
```

```python
tscv = TimeSeriesSplit(n_splits=5)

print("Time-series cross-validation created.")
```

```python
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'min_samples_leaf': [5, 10, 20]
}
```

```python
grid_search = GridSearchCV(
    RandomForestClassifier(
        random_state=42
    ),
    param_grid,
    cv=tscv,
    scoring='accuracy',
    n_jobs=-1
)

grid_search.fit(
    X_train,
    y_train
)
```

```python
print(
    "Best parameters:",
    grid_search.best_params_
)

print(
    "Best CV accuracy:",
    round(
        grid_search.best_score_ * 100,
        2
    ),
    "%"
)
```

```python
best_rf = grid_search.best_estimator_

y_pred_best_rf = best_rf.predict(X_test)

tuned_accuracy = accuracy_score(
    y_test,
    y_pred_best_rf
)

print(
    "Tuned Random Forest Test Accuracy:",
    round(
        tuned_accuracy * 100,
        2
    ),
    "%"
)
```

```python
print(
    classification_report(
        y_test,
        y_pred_best_rf
    )
)
```

```python
final_results = pd.DataFrame({
    'Model': [
        'Majority Baseline',
        'Logistic Regression',
        'Random Forest',
        'XGBoost',
        'Tuned Random Forest'
    ],
    'Accuracy (%)': [
        majority_baseline * 100,
        lr_accuracy * 100,
        rf_accuracy * 100,
        xgb_accuracy * 100,
        tuned_accuracy * 100
    ]
})

final_results['Accuracy (%)'] = (
    final_results['Accuracy (%)'].round(2)
)

display(final_results)
```

```python
test_df = ml_df.iloc[
    split_idx:
].copy().reset_index(drop=True)

test_df['pred'] = y_pred_best_rf

display(
    test_df[
        ['date', 'close', 'next_day_target', 'pred']
    ].head()
)
```

```python
test_df['next_day_return_pct'] = (
    (
        test_df['close'].shift(-1)
        - test_df['close']
    )
    / test_df['close']
) * 100
```

```python
import numpy as np

test_df['strategy_return'] = np.where(
    test_df['pred'] == 1,
    test_df['next_day_return_pct'],
    0
)
```

```python
test_df['cumulative_strategy'] = (
    1 +
    test_df['strategy_return']
    .fillna(0) / 100
).cumprod()

test_df['cumulative_market'] = (
    1 +
    test_df['next_day_return_pct']
    .fillna(0) / 100
).cumprod()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    test_df['date'],
    test_df['cumulative_strategy'],
    label='Model Strategy'
)

plt.plot(
    test_df['date'],
    test_df['cumulative_market'],
    label='Buy & Hold NIFTY50'
)

plt.xlabel('Date')
plt.ylabel('Growth of ₹1')
plt.title(
    'Strategy vs Buy-and-Hold NIFTY50'
)

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
strategy_final = (
    test_df['cumulative_strategy'].iloc[-2]
)

market_final = (
    test_df['cumulative_market'].iloc[-2]
)

print(
    "Model Strategy Growth of ₹1:",
    round(strategy_final, 4)
)

print(
    "Buy & Hold Growth of ₹1:",
    round(market_final, 4)
)
```

```python
print(
    "Model Strategy Return:",
    round(
        (strategy_final - 1) * 100,
        2
    ),
    "%"
)

print(
    "Buy & Hold Return:",
    round(
        (market_final - 1) * 100,
        2
    ),
    "%"
)
```
